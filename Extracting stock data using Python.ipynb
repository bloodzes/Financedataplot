```python
!pip install yfinance
!pip install matplotlib
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

tsla = yf.Ticker("TSLA")

#history method to download historical data for stock
# period can be in 1d, 1mo, 2y, ytd, or max
tslashare_pricedata = tsla.history(period='max')

#display downloaded data
tslashare_pricedata.head()

tslashare_pricedata.reset_index(inplace=True)
tslashare_pricedata.plot(x='Date',y='Open')

plt.title("TSLA Historical Open Price")
plt.show()
```

```python
tsla.dividends
tsla.dividends.plot(x = 'Date', y='Dividends US$')
plt.title('Tesla Dividends Paid')
plt.show()
```
